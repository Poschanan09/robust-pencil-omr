## 3. Histogram Analysis: Shadowed vs Uniform Lighting

```python
canvas = np.ones((600, 400), dtype=np.uint8) * 240
gradient = np.tile(np.linspace(0, 160, 400), (600, 1)).astype(np.uint8)
shadowed = cv2.subtract(canvas, gradient)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].imshow(canvas, cmap='gray', vmin=0, vmax=255)
axes[0].set_title('Uniform Canvas')
axes[1].imshow(shadowed, cmap='gray', vmin=0, vmax=255)
axes[1].set_title('Shadowed Capture')
plt.tight_layout()
plt.show()
```
