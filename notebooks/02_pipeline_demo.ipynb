Interactive step-by-step execution of the 7-step classical DIP pipeline:
1. Input Quality Check (Laplacian Blur & Histogram Exposure)
2. Document Boundary Detection (Canny + Douglas-Peucker)
3. Homography Perspective Rectification (Projective 4-Point Warp)
4. Local Adaptive Gaussian Thresholding + Mathematical Morphology
5. Inner-Ring Masking & Feature Extraction
6. Tri-State Decision Logic (FILLED, BLANK, AMBIGUOUS)
7. Diagnostic Overlay Rendering

```python
import sys
sys.path.append('..')
from pathlib import Path
import cv2
import numpy as np
import matplotlib.pyplot as plt

from omr.pipeline import OMRPipeline
from omr.decision_engine import TriStateDecisionEngine
```

## 1. Initialize OMR Pipeline

```python
pipeline = OMRPipeline()
print('Pipeline successfully initialized with canonical DPI 300!')
```

## 2. Run Pipeline on Template Sample

```python
sample_path = Path('../templates/answer_sheet_a4.png')
if sample_path.exists():
    results = pipeline.process_single_image(sample_path)
    print(f"Processed Status: {results['status']}")
    print('Summary:', results['summary'])
else:
    print('Please provide an answer sheet image to test.')
```

## 3. Visualize Intermediate Pipeline Stages

```python
if sample_path.exists() and 'debug_images' in results:
    debugs = results['debug_images']
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    if 'canny_edges' in debugs:
        axes[0].imshow(debugs['canny_edges'], cmap='gray')
        axes[0].set_title('Step 2: Canny Edges')
    if 'binary_clean' in debugs:
        axes[1].imshow(debugs['binary_clean'], cmap='gray')
        axes[1].set_title('Step 4: Binary Clean')
    if 'overlay' in debugs:
        axes[2].imshow(cv2.cvtColor(debugs['overlay'], cv2.COLOR_BGR2RGB))
        axes[2].set_title('Step 7: Diagnostic Overlay')
    plt.tight_layout()
    plt.show()
```
